# Remotion HTML Embed Test

下面是一个文本 block。下一格会通过 `IPython.display.HTML` 注入一个 HTML 片段，
片段里使用 **Remotion Player** 播放一个简单动画（基于 `useCurrentFrame()`）。



```python
from IPython.display import HTML, display

html = """
<div style="font-family: ui-sans-serif, system-ui; margin: 12px 0 8px; font-size: 14px; color: #111;">
  Remotion embed preview (iframe)
</div>
<iframe
  style="width: 100%; max-width: 960px; height: 440px; border: 1px solid #ddd; border-radius: 10px; background: #0f172a;"
  sandbox="allow-scripts allow-same-origin"
  srcdoc='
<!doctype html>
<html>
  <head>
    <meta charset="utf-8" />
    <style>
      html, body, #root { margin: 0; width: 100%; height: 100%; background: #0f172a; color: #fff; }
      .err { white-space: pre-wrap; font: 13px/1.5 ui-monospace, SFMono-Regular, Menlo, monospace; padding: 12px; color: #fecaca; }
    </style>
  </head>
  <body>
    <div id="root"></div>
    <script type="module">
      const rootEl = document.getElementById("root");
      const showErr = (e) => {
        const msg = (e && e.stack) ? e.stack : String(e);
        rootEl.innerHTML = `<div class="err">Remotion load/render failed:\n${msg}</div>`;
      };

      window.addEventListener("error", (ev) => showErr(ev.error || ev.message));
      window.addEventListener("unhandledrejection", (ev) => showErr(ev.reason || ev));

      try {
        const reactV = "18.3.1";
        const [ReactMod, ReactDOMMod, PlayerMod, RemotionMod] = await Promise.all([
          import(`https://esm.sh/react@${reactV}`),
          import(`https://esm.sh/react-dom@${reactV}/client`),
          import(`https://esm.sh/@remotion/player@4.0.366?deps=react@${reactV},react-dom@${reactV}`),
          import(`https://esm.sh/remotion@4.0.366?deps=react@${reactV},react-dom@${reactV}`),
        ]);

        const React = ReactMod.default ?? ReactMod;
        const {createRoot} = ReactDOMMod;
        const {Player} = PlayerMod;
        const {AbsoluteFill, interpolate, spring, useCurrentFrame, useVideoConfig} = RemotionMod;

        const Demo = () => {
          const frame = useCurrentFrame();
          const {fps} = useVideoConfig();

          const moveX = interpolate(frame, [0, 3 * fps], [40, 760], {extrapolateRight: "clamp"});
          const scale = spring({frame, fps, config: {damping: 12, stiffness: 120}});
          const rotate = interpolate(frame, [0, 3 * fps], [0, 360]);

          return React.createElement(
            AbsoluteFill,
            {style: {background: "linear-gradient(135deg, #0f172a, #1e293b)", color: "#fff", fontFamily: "ui-sans-serif, system-ui"}},
            React.createElement("div", {style: {position: "absolute", left: 24, top: 20, fontSize: 22, fontWeight: 700}}, "Remotion Animation"),
            React.createElement("div", {
              style: {
                position: "absolute",
                top: 190,
                left: moveX,
                width: 72,
                height: 72,
                borderRadius: 16,
                background: "#22d3ee",
                transform: `scale(${0.6 + 0.4 * scale}) rotate(${rotate}deg)`,
                boxShadow: "0 12px 24px rgba(34,211,238,0.35)",
              }
            }),
            React.createElement("div", {style: {position: "absolute", left: 24, bottom: 20, fontSize: 14, opacity: 0.85}}, "Rendered in notebook via HTML snippet + Remotion Player")
          );
        };

        const root = createRoot(rootEl);
        root.render(
          React.createElement(Player, {
            component: Demo,
            durationInFrames: 180,
            fps: 30,
            compositionWidth: 960,
            compositionHeight: 400,
            controls: true,
            autoPlay: true,
            loop: true,
            style: {width: "100%", height: "100%"}
          })
        );
      } catch (e) {
        showErr(e);
      }
    </script>
  </body>
</html>
'
></iframe>
"""

display(HTML(html))
```
