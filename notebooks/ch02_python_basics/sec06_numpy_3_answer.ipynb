# 思考题：极简版的卷积操作

在图像处理中，卷积操作 `convolution` 指的是：

- 用一个固定大小的 `kernel`
- 与图像中的每一个局部 `patch` 做内积
- 生成一张新的图像。

现在，我们尝试实现一个极简版的卷积操作，让我们假设：

- 输入图像是个二维数组，即黑白图像
- `kernel` 的尺寸是 `2×2`
- 卷积的 `stride` 为 `2`

**提示:**
- 只使用 `reshape`、`transpose` 和 `矩阵乘法`
- Kernel 定义: $\begin{bmatrix} 0.75 & 0.75 \\ -0.75 & -0.75 \end{bmatrix}$

# 答案

```python
# @title 辅助函数
import numpy as np
from PIL import Image
from IPython.display import display

!wget -q -nc --directory-prefix=downloads https://raw.githubusercontent.com/WuDianQiBian/ml-course/master/notebooks/ch02_python_basics/assets/Clark.jpg

def read_image(path="downloads/Clark.jpg"):
    return np.asarray(Image.open(path)).copy()

def display_image(img):
    if img.dtype != np.uint8:
        img = np.clip(img, 0, 255).astype(np.uint8)
    display(Image.fromarray(img))
```

```python
# 读取图片，转为灰度图
img = read_image().astype(np.float32)
gray = np.mean(img, axis=2)  # (H, W)

H, W = gray.shape

print(f"{gray.shape=}")
display_image(gray.astype(np.uint8))
```

```python
# 定义 2x2 kernel（提取横向边缘）
kernel = np.array([
    [ 0.75,  0.75],
    [-0.75, -0.75]
], dtype=np.float32)
```

```python
# Step 1: reshape，把图像分成"互不重叠"的 2x2 的 小块 (patches)
# (H, W) -> (H//2, 2, W//2, 2)
patches = gray.reshape(H // 2,
                       2,
                       W // 2,
                       2)

# Step 2: transpose，把两个 "2" 放在一起   👇
# (H//2, 2, W//2, 2) -> (H//2, W//2,   2, 2)
patches = np.transpose(patches, [0, 2, 1, 3])

# Step 3: 拍平每个 patch 为长度 4 的向量
# (H//2, W//2, 2, 2) -> (H//2, W//2, 4)
patches = patches.reshape(H // 2, W // 2, 4)

# Step 4: kernel 也拍平为 (1, 4, 1)
k = kernel.reshape(1, 4, 1)

# Step 5: 矩阵乘法，每个 patch 和 kernel 做内积
# (H//2, W//2, 4) @ (1, 4, 1) -> (H//2, W//2, 1)
output = patches @ k

# Step 6: 去掉最后一维
output = output.reshape(H // 2, W // 2)
print(f"{output.shape=}")
```

```python
# 展示卷积结果
display_image(np.abs(output).clip(0, 255).astype(np.uint8))
```
